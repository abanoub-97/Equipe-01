```python
import pandas as pd
import random
from datetime import datetime, timedelta

# Configuration reflecting Givaudan's actual divisions and categories
divisions = {
    'Taste & Wellbeing': ['Sweet Goods', 'Dairy', 'Beverages', 'Savoury'],
    'Fragrance & Beauty': ['Fine Fragrances', 'Consumer Products', 'Fragrance Ingredients', 'Active Beauty']
}

regions = ['North America', 'Europe', 'Latin America', 'Asia-Pacific', 'Middle East & Africa']
roles = ['Decision-maker', 'Technical', 'Marketing', 'Purchasing', 'Regulatory']
stages = ['Lead', 'Brief Received', 'Evaluation/Sampling', 'Negotiation', 'Closed Won', 'Closed Lost']
num_clients = 50

data = []
current_time = datetime.now()

for i in range(num_clients):
    client_id = f"Client_{random.randint(1000, 9999)}"
    division = random.choice(list(divisions.keys()))
    category = random.choice(divisions[division])
    region = random.choice(regions)
    
    # Customer history profile
    is_returning_customer = random.choice([True, False])
    avg_order_volume_kg = random.randint(100, 5000)
    order_frequency_months = random.randint(1, 12)
    
    interactions_count = random.randint(5, 15) # Ensuring >= 5 interactions per client
    
    for j in range(interactions_count):
        # Ensuring strictly chronological order with realistic timestamps
        days_ago = random.randint(0, 730) # Within last 2 years
        hours_ago = random.randint(1, 24)
        interaction_time = current_time - timedelta(days=days_ago, hours=hours_ago)
        
        stage = random.choice(stages)
        role = random.choice(roles)
        proof_strength = random.choice(['+', '++', '+++'])
        
        # Mocking transcript with PII placeholder and specific business context
        transcript = f"Client from {region} ({category} sector) discussed project requirements. [PII REMOVED]. Feedback provided on formulation stability and pricing competitiveness."
        
        data.append({
            'client_id': client_id,
            'division': division,
            'category': category,
            'region': region,
            'is_returning_customer': is_returning_customer,
            'avg_order_volume_kg': avg_order_volume_kg,
            'order_frequency_months': order_frequency_months,
            'interaction_timestamp': interaction_time,
            'pipeline_stage': stage,
            'stakeholder_role': role,
            'transcript': transcript,
            'proof_strength': proof_strength,
            'language': random.choice(['EN', 'FR']) # Reflecting global operations
        })

df = pd.DataFrame(data)

# Sorting by client and timestamp to strictly enable recency bias and historical tracking
df = df.sort_values(by=['client_id', 'interaction_timestamp']).reset_index(drop=True)

# Displaying info and head to verify structure
print(f"Total Interactions Generated: {len(df)}")
print("\nSample Data Structure:")
print(df[['client_id', 'division', 'category', 'pipeline_stage', 'interaction_timestamp']].head(10))
```

```python
df.to_csv('givaudan_crm_data.csv', index=False)
```
